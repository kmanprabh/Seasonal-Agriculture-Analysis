```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("/content/seasonal_agriculture_performance_dataset.csv")
```

```python
df.head()
```

```python
df.columns
```

```python
df.describe().T
```

```python
df["Season"].value_counts()
df['Crop'].unique()
```

```python
df.isnull().sum()
```

```python
missing = df.isnull().sum()

missing = missing[missing > 0]

missing.plot(kind='bar')

plt.title("Missing Values in Dataset")
plt.xlabel("Columns")
plt.ylabel("Number of Missing Values")
plt.show()
```

```python
df['Rainfall_mm'].fillna(df['Rainfall_mm'].median(), inplace=True)

df['Soil_Moisture_pct'].fillna(df['Soil_Moisture_pct'].median(),inplace=True)

df['Yield_Tonnes_Ha'].fillna(df['Yield_Tonnes_Ha'].median(),inplace=True)
```

```python
df.isnull().sum()
```

```python
df.duplicated().sum()
```

HOW MUCH DATA BELONG TO EACH SEASON


```python
season_counts = df['Season'].value_counts()

season_counts
```

```python
plt.figure(figsize=(8,5))

sns.countplot(
    data=df,
    x='Season',
    order=df['Season'].value_counts().index
)

plt.title("Distribution of Agricultural Records by Season")
plt.xlabel("Season")
plt.ylabel("Number of Farms")
plt.show()
```

AVERAGE YEILD BY SEASON


```python
season_yield = df.groupby('Season')['Yield_Tonnes_Ha'].mean().sort_values(ascending=False)

season_yield
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Season',
    y='Yield_Tonnes_Ha',
    estimator='mean'
)

plt.title("Average Crop Yield by Season")
plt.ylabel("Average Yield (Tonnes/Ha)")
plt.xlabel("Season")
plt.show()
```

PRODUCTION BY SEASON

```python
season_production = df.groupby('Season')['Production_Tonnes'].mean()

season_production
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Season',
    y='Production_Tonnes'
)

plt.title("Average Agricultural Production by Season")
plt.ylabel("Production (Tonnes)")
plt.show()
```

REVENUE BY SEASON

```python
season_revenue = df.groupby('Season')['Revenue_INR'].mean()

season_revenue
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Season',
    y='Revenue_INR'
)

plt.title("Average Revenue by Season")
plt.ylabel("Average Revenue (INR)")
plt.show()
```

PROFIT BY SEASON

```python
season_profit = df.groupby('Season')['Profit_INR'].mean()

season_profit
```

```python
plt.figure(figsize=(8,5))

sns.barplot(
    data=df,
    x='Season',
    y='Profit_INR'
)

plt.title("Average Profit by Season")
plt.ylabel("Average Profit (INR)")
plt.show()
```

MOST PROFITABLE CROPS

```python
crop_profit = (
    df.groupby('Crop')['Profit_INR']
    .mean()
    .sort_values(ascending=False)
)

crop_profit
```

```python
plt.figure(figsize=(10,6))

sns.barplot(
    x=crop_profit.values,
    y=crop_profit.index
)

plt.title("Average Profit by Crop")
plt.xlabel("Average Profit (INR)")
plt.ylabel("Crop")
plt.show()
```
